Then explain which conditional independence assumption [[Naive Bayes]] makes and why it is "naive".

*Your answer:* ✏️ (double-click to write here)

<details>
<summary><b>💡 Hint</b></summary>

Think of a common cause: $Z$ influences both $X$ and $Y$, but once $Z$ is known, $X$ tells you nothing more about $Y$.

</details>

<details>
<summary><b>✅ Solution</b></summary>

often still ranks classes well.

### Exercise 2 · MLE, MAP and regularisation
*🧠 Concept · ★★☆*

Write the MAP objective for linear regression with Gaussian noise $\mathcal N(0,\sigma^2)$ and prior $w\sim\mathcal N(0,\tau^2 I)$, and
show that it is ridge regression. Which $\lambda$ do you get? What changes with a Laplace prior? Why does MAP converge to MLE as $N\to\infty$?

*Your answer:* ✏️ (double-click to write here)

<details>
<summary><b>💡 Hint</b></summary>

MAP maximises $\log p(D\mid w) + \log p(w)$. The Gaussian log-prior is $-\frac{1}{2\tau^2}\lVert w\rVert^2 + \text{const}$.

</details>

<details>
<summary><b>✅ Solution</b></summary>

$-\log p(w\mid D) = \frac{1}{2\sigma^2}\sum_i (y_i-w^\top x_i)^2 + \frac{1}{2\tau^2}\lVert w\rVert^2 + \text{const}$.
Multiplying by $2\sigma^2$: minimise $\lVert Xw-y\rVert^2 + \lambda\lVert w\rVert^2$ with $\lambda=\sigma^2/\tau^2$ (ridge).
A Laplace prior $p(w)\propto e^{-\lVert w\rVert_1/b}$ gives an L1 penalty (Lasso). The likelihood term grows with $N$ while the
prior term stays fixed, so for large $N$ the prior's influence vanishes and MAP → MLE. See [[Overfitting and Regularization]].

</details>

### Exercise 3 · LLN, CLT and Hoeffding
*🧠 Concept · ★☆☆*

In one or two sentences each, state what the law of large numbers, the central limit theorem and Hoeffding's inequality say
about the sample mean $\bar X_n$ of i.i.d. variables. Which one gives a non-asymptotic guarantee, and why is that what PAC
learning needs?

*Your answer:* ✏️ (double-click to write here)

<details>
<summary><b>💡 Hint</b></summary>

LLN: *where* $\bar X_n$ goes. CLT: the *shape* of its fluctuations. Hoeffding: a *finite-$n$ bound* on the tail.

</details>

<details>
<summary><b>✅ Solution</b></summary>

**LLN**: $\bar X_n \to \mu$ (in probability / almost surely) as $n\to\infty$.
**CLT**: $\sqrt n(\bar X_n - \mu)/\sigma \to \mathcal N(0,1)$, so fluctuations are about $\sigma/\sqrt n$ and approximately Gaussian.
**Hoeffding**: for $X_i\in[a,b]$, $P(|\bar X_n-\mu|\ge\varepsilon)\le 2\exp\!\left(-\frac{2n\varepsilon^2}{(b-a)^2}\right)$ for **every** $n$.
Only Hoeffding is non-asymptotic; PAC bounds need "with probability $\ge 1-\delta$ the training error is within $\varepsilon$ of the
true error" for a *finite* training set, which is exactly that form.

</details>

### Exercise 4 · Conjugate priors
*🧠 Concept · ★★☆*

What does it mean for the Beta distribution to be *conjugate* to the Bernoulli likelihood? Interpret the hyperparameters
$\alpha,\beta$ of a $\mathrm{Beta}(\alpha,\beta)$ prior as "pseudo-counts". What is the multi-class analogue?

*Your answer:* ✏️ (double-click to write here)

<details>
<summary><b>💡 Hint</b></summary>

Multiply $\theta^{\alpha-1}(1-\theta)^{\beta-1}$ by the likelihood $\theta^k(1-\theta)^{n-k}$.

</details>

<details>
<summary><b>✅ Solution</b></summary>

Conjugate: the posterior is in the same family as the prior. $p(\theta\mid D)\propto\theta^{k+\alpha-1}(1-\theta)^{n-k+\beta-1}$, i.e.
$\mathrm{Beta}(\alpha+k,\ \beta+n-k)$. So $\alpha-1$ and $\beta-1$ behave like previously seen heads and tails (and $\alpha+\beta$
measures how strongly you believe the prior). The multi-class analogue is the Dirichlet prior for the Categorical/Multinomial;
with all $\alpha_k=2$ the MAP estimate is Laplace ("add-one") smoothing used in Naive Bayes.

</details>

<details>
<summary><b>💡 Hint</b></summary>

$\ell = -\frac n2\log(2\pi\sigma^2) - \frac{1}{2\sigma^2}\sum(x_i-\mu)^2$. Differentiate w.r.t. $\mu$, then w.r.t. $\sigma^2$.

</details>

    d = len(mu)
    L = np.linalg.cholesky(Sigma)
    Z = np.linalg.solve(L, (X - mu).T)          # d x N

````python
flips = (np.random.default_rng(3).random(200) < 0.3).astype(int)
post_ab = None
post_mean = None
cred_int = None

_k = flips.sum(); _ref = stats.beta(1 + _k, 1 + 200 - _k)
check('posterior (alpha, beta)', post_ab, [1 + _k, 201 - _k])
check('posterior mean', post_mean, _ref.mean())
check('95% interval', cred_int, _ref.interval(0.95))
````

<details>
<summary><b>💡 Hint</b></summary>

Each head adds 1 to $\alpha$, each tail adds 1 to $\beta$; the order of the flips does not matter.

</details>

<details>
<summary><b>✅ Solution</b></summary>

Updating one flip at a time gives the same posterior as one batch update: yesterday's posterior is today's prior. The interval narrows roughly like $1/\sqrt n$ around the true $0.3$.

```python
import matplotlib.pyplot as plt
flips = (np.random.default_rng(3).random(200) < 0.3).astype(int)
a, b = 1, 1
grid = np.linspace(0, 1, 300)
for t, f in enumerate(flips, 1):
    a, b = a + f, b + (1 - f)
    if t in (10, 50, 200):
        plt.plot(grid, stats.beta.pdf(grid, a, b), label=f'{t} flips')
plt.plot(grid, stats.beta.pdf(grid, 1, 1), label='prior'); plt.legend(); plt.show()
post_ab = [a, b]
post_mean = a / (a + b)
cred_int = stats.beta.interval(0.95, a, b)
```

### Exercise 17 · Checking Hoeffding empirically
*💻 Code · ★★★*

For Bernoulli(0.5) variables, $\varepsilon=0.1$ and $n\in\{10, 50, 100, 200\}$, estimate $P(|\bar X_n - 0.5|\ge\varepsilon)$ with
$20\,000$ repetitions each and compare with the Hoeffding bound $2e^{-2n\varepsilon^2}$. Store the empirical probabilities in
`emp_probs` and the bounds in `hoeff_bounds` (lists in the order of `n_list`). Is the bound tight?

````python
n_list = [10, 50, 100, 200]
emp_probs = None
hoeff_bounds = None

check('bounds', hoeff_bounds, [2 * np.exp(-2 * n * 0.01) for n in n_list])
check('empirical prob never exceeds bound', None if emp_probs is None else all(e <= b for e, b in zip(emp_probs, hoeff_bounds)), True)
check('empirical probs match the exact binomial tail', emp_probs,
      [stats.binom.cdf(np.floor(0.4 * n + 1e-9), n, 0.5) + stats.binom.sf(np.ceil(0.6 * n - 1e-9) - 1, n, 0.5) for n in n_list], tol=0.015)
````

<details>
<summary><b>💡 Hint</b></summary>

`means = rng.binomial(n, 0.5, size=20000) / n`, then `np.mean(np.abs(means - 0.5) >= 0.1 - 1e-12)`.

</details>

<details>
<summary><b>✅ Solution</b></summary>

The bound always holds but is loose (e.g. for $n=10$ it is $1.64$, a vacuous bound, while the true probability is about $0.75$). Hoeffding only uses the range $[0,1]$, not the variance; Bernstein-type or CLT approximations are tighter but either need more assumptions or are only asymptotic.

```python
n_list = [10, 50, 100, 200]
r_h = np.random.default_rng(7)
emp_probs = [np.mean(np.abs(r_h.binomial(n, 0.5, size=20000) / n - 0.5) >= 0.1 - 1e-12) for n in n_list]
hoeff_bounds = [2 * np.exp(-2 * n * 0.1 ** 2) for n in n_list]
print(np.round(emp_probs, 4), np.round(hoeff_bounds, 4))
```

</details>
