# Twenty Questions: Danish animals 🦔
Think of a Danish animal, and the notebook guesses it by asking yes/no questions. The trick is to always ask the question that splits the remaining candidates **most evenly** (maximum entropy, i.e. maximum information gain). That's exactly how a decision tree chooses splits, and how *Akinator* works.

```python
def split_entropy(items, q):
    n = len(items)
    if n == 0:
        return 0.0
    yes = sum(TABLE[a][q] for a in items)
    p = yes / n
    if p in (0, 1):
        return 0.0
    return float(-(p * np.log2(p) + (1 - p) * np.log2(1 - p)))
```

```python
def best_question(items, questions):
    scores = {q: split_entropy(items, q) for q in questions}
    return max(scores, key=scores.get)
```
